```python

import requests
import pdfplumber
from pathlib import Path

URL_PDF = (
    "https://firebasestorage.googleapis.com/v0/b/"
    "instad-dj.appspot.com/o/ICA%20Service%2F2025%2F"
    "ICA_SERV_T12025.pdf?alt=media&token="
    "9557bdf4-be3c-40ef-b122-e3b110d8d1c1"
)

DOSSIER = Path("/content/ica_services")
DOSSIER.mkdir(parents=True, exist_ok=True)

PDF_PATH = DOSSIER / "ICA_SERV_T12025.pdf"

reponse = requests.get(URL_PDF, timeout=60)
reponse.raise_for_status()

if not reponse.content.startswith(b"%PDF"):
    raise ValueError("Le fichier téléchargé n'est pas un PDF valide.")

PDF_PATH.write_bytes(reponse.content)

pages = []

with pdfplumber.open(PDF_PATH) as pdf:
    print("Nombre de pages :", len(pdf.pages))

    for numero, page in enumerate(pdf.pages, start=1):
        texte = page.extract_text() or ""
        pages.append(texte)
        print(f"Page {numero} : {len(texte)} caractères")

texte_complet = "\n".join(pages)

(DOSSIER / "texte_extrait.txt").write_text(
    texte_complet, encoding="utf-8"
)

print(texte_complet[:5000])
```

```python

%%writefile /content/ica_services/app.py
from pathlib import Path

import pandas as pd
import plotly.express as px
import streamlit as st

st.set_page_config(
    page_title="ICA Services | Djibouti",
    page_icon="📊",
    layout="wide",
)

st.title("📊 Tableau de bord public — ICA Services")
st.subheader("République de Djibouti | T1 2025")
st.caption(
    "Source : INSTAD, Bulletin de l'Indice du chiffre "
    "d'affaires dans les services, premier trimestre 2025. "
    "Base 2022."
)

# Valeurs reprises du tableau statistique de la page 9 du PDF.
# Les deux variations sont conservées séparément.
donnees = [
    ("Ensemble des services", 136.5, -7.7, -2.6),
    ("Transports et entreposage", 156.4, -1.1, 2.4),
    ("Transports terrestres", 115.4, 1.6, -12.8),
    ("Entreposage et auxiliaires de transport", 161.1, -1.3, 3.9),
    ("Hébergement et restauration", 106.9, -21.0, -16.9),
    ("Hébergement", 66.0, -31.3, -35.3),
    ("Restauration et débits de boisson", 135.3, -16.7, -8.0),
    ("Information et communication", 101.0, -47.9, 1.6),
    ("Télécommunications", 100.8, -48.1, 1.7),
    ("Activités informatiques", 124.0, 1.4, -6.5),
    ("Activités immobilières", 73.2, 22.9, -23.9),
    ("Activités spécialisées, scientifiques et techniques",
     38.8, 4.1, -39.6),
    ("Services de soutien et de bureau", 106.7, 3.5, -8.4),
    ("Location et location-bail", 66.8, 5.6, -30.8),
    ("Ressources humaines", 110.3, 0.9, 0.2),
    ("Agences de réservation et voyagistes", 102.4, 7.9, 80.1),
    ("Enquêtes et sécurité", 146.2, 39.8, -32.7),
    ("Soutien aux bâtiments et aménagement paysager",
     124.1, -7.6, -49.1),
    ("Enseignement", 154.8, -29.4, -25.0),
    ("Santé humaine et action sociale", 115.4, 11.0, -0.4),
    ("Activités artistiques, sportives et récréatives",
     158.7, 17.6, 47.2),
    ("Autres activités de services marchands", 90.7, -20.7, -4.6),
]

df = pd.DataFrame(
    donnees,
    columns=[
        "Secteur",
        "Indice T1 2025",
        "Variation trimestrielle (%)",
        "Variation annuelle (%)",
    ],
)

st.sidebar.header("Filtres")
secteurs = st.sidebar.multiselect(
    "Choisir les secteurs",
    options=df["Secteur"].tolist(),
    default=df["Secteur"].tolist(),
)
indicateur = st.sidebar.selectbox(
    "Indicateur à comparer",
    [
        "Variation trimestrielle (%)",
        "Variation annuelle (%)",
        "Indice T1 2025",
    ],
)

vue = df[df["Secteur"].isin(secteurs)].copy()

if vue.empty:
    st.warning("Sélectionnez au moins un secteur.")
    st.stop()

ensemble = df.loc[df["Secteur"] == "Ensemble des services"].iloc[0]

c1, c2, c3, c4 = st.columns(4)
c1.metric("Indice global T1 2025", f"{ensemble['Indice T1 2025']:.1f}")
c2.metric("Variation trimestrielle", f"{ensemble['Variation trimestrielle (%)']:+.1f} %")
c3.metric("Variation annuelle", f"{ensemble['Variation annuelle (%)']:+.1f} %")
c4.metric("Secteurs affichés", str(len(vue)))

st.divider()
st.subheader("Comparaison par secteur")

vue_graphique = vue.sort_values(indicateur)
fig = px.bar(
    vue_graphique,
    x=indicateur,
    y="Secteur",
    orientation="h",
    color=indicateur,
    color_continuous_scale="RdYlGn",
    title=f"ICA Services — {indicateur}",
    hover_data={
        "Indice T1 2025": ":.1f",
        "Variation trimestrielle (%)": ":.1f",
        "Variation annuelle (%)": ":.1f",
    },
)
if "Variation" in indicateur:
    fig.add_vline(x=0, line_dash="dash", line_color="gray")
fig.update_layout(
    height=max(500, len(vue_graphique) * 28),
    yaxis_title="",
    xaxis_title=indicateur,
    coloraxis_showscale=False,
)
st.plotly_chart(fig, use_container_width=True)

st.subheader("Tableau statistique")
st.dataframe(
    vue,
    use_container_width=True,
    hide_index=True,
    column_config={
        "Indice T1 2025": st.column_config.NumberColumn(
            format="%.1f"
        ),
        "Variation trimestrielle (%)":
            st.column_config.NumberColumn(format="%.1f %%"),
        "Variation annuelle (%)":
            st.column_config.NumberColumn(format="%.1f %%"),
    },
)

csv = vue.to_csv(index=False).encode("utf-8-sig")
st.download_button(
    "Télécharger les données CSV",
    data=csv,
    file_name="ica_services_t1_2025.csv",
    mime="text/csv",
)

st.divider()
st.subheader("Méthodologie")
st.write(
    "L'indice mesure l'évolution du chiffre d'affaires des "
    "entreprises de services couvertes par l'enquête. "
    "L'année de base est 2022. Les variations trimestrielles "
    "comparent T1 2025 à T4 2024 ; les variations annuelles "
    "comparent T1 2025 à T1 2024."
)
st.caption(
    "Source : INSTAD, Bulletin ICA Services T1 2025, tableau 1. "
    "Les chiffres sont reproduits à des fins de consultation."
)
```
